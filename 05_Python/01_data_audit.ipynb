```python
order_counts = orders.groupby("Customer ID")["First order vs repeat"].value_counts().unstack(fill_value=0)
```

```python
check = customer[["Customer ID", "Total orders", "RePurchased"]].merge(
    order_counts,
    on="Customer ID",
    how="left"
)
```

```python
check["orders_sum"] = check["First"] + check["Repeat"]

# Contradiction A: total orders in customers.csv doesn't match count of orders in orders.csv
contradiction_a = check[check["Total orders"] != check["orders_sum"]]

# Contradiction B: RePurchased = 'Y' but no orders marked 'Repeat'
contradiction_b = check[(check["RePurchased"] == "Y") & (check["Repeat"] == 0)]

# Contradiction C: Total orders > 1 but zero orders marked 'Repeat'
contradiction_c = check[(check["Total orders"] > 1) & (check["Repeat"] == 0)]

# Contradiction D: Total orders > 1 but RePurchased is No'
contradiction_d = check[(check["Total orders"] > 1) & (check["RePurchased"] == "N")]

print("Contradiction A (order count mismatch):", len(contradiction_a))
print("Contradiction B (RePurchased=Y but no Repeat orders):", len(contradiction_b))
print("Contradiction C (>1 order but no Repeat orders):", len(contradiction_c))
print("Total orders > 1 but RePurchased is No:",len(contradiction_d))
```

```python
set(contradiction_b["Customer ID"]) == set(contradiction_c["Customer ID"])
```

```python
suspect_ids = contradiction_b["Customer ID"].tolist()
orders[orders["Customer ID"].isin(suspect_ids)].sort_values(["Customer ID", "Order date & time"])
```

```python
orders["order_rank"] = orders.groupby("Customer ID")["Order date & time"].rank(method="first")
orders["Corrected order type"] = orders["order_rank"].apply(lambda r: "First" if r == 1 else "Repeat")
```

```python
mismatches = orders[orders["First order vs repeat"] != orders["Corrected order type"]]
print("Total mismatched rows across entire dataset:", len(mismatches))
print("Unique customers affected:", mismatches["Customer ID"].nunique())
```

```python
set(suspect_ids).issubset(set(mismatches["Customer ID"]))
```

```python
mismatches.groupby("Customer ID")["Order date & time"].agg(lambda x: x.max() - x.min()).describe()
```

```python
affected_ids = mismatches["Customer ID"].unique()
gaps = orders[orders["Customer ID"].isin(affected_ids)].groupby("Customer ID")["Order date & time"].agg(lambda x: x.max() - x.min())
gaps.describe()
```

```python
large_gap_ids = gaps[gaps > pd.Timedelta(days=30)].index
orders[orders["Customer ID"].isin(large_gap_ids)].sort_values(["Customer ID", "Order date & time"])
```

## Finding: `First order vs repeat` mislabeled for 30 customers (~0.18%)
Comparing the original label against a date-derived version (earliest order per customer = First, rest = Repeat) found 30 customers with at least one wrong label. Root cause unclear — roughly half show orders placed within 1-2 days of each other (likely a pipeline timing issue), the rest show gaps of months to years (possibly a reactivation-relabeling rule, unconfirmed). Given the small scale (<0.2% of customers), the original column is kept as-is for traceability, but a derived `Corrected order type` column is used for all repeat-purchase analysis instead.
